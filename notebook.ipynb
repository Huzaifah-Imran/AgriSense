```python
import subprocess
import time
import os
import requests
import pinggy

# --------------------------------------------------
# 1. Start Ollama
# --------------------------------------------------

env = os.environ.copy()

env["OLLAMA_HOST"] = "0.0.0.0:11434"
env["OLLAMA_ORIGINS"] = "*"
env["OLLAMA_KEEP_ALIVE"] = "-1" 

ollama_process = subprocess.Popen(
    ["ollama", "serve"],
    env=env,
)

print("Starting Ollama...")

# --------------------------------------------------
# 2. Wait until Ollama is actually ready
# --------------------------------------------------

ollama_url = "http://127.0.0.1:11434"

for i in range(60):

    try:
        response = requests.get(
            f"{ollama_url}/api/tags",
            timeout=2
        )

        if response.status_code == 200:

            print("Ollama is ready!")
            print(response.text)

            break

    except requests.exceptions.RequestException:
        pass

    time.sleep(1)

else:

    print("Ollama failed to start.")

    raise RuntimeError(
        "Ollama did not become ready."
    )

```

```python
# --------------------------------------------------
# 3. Start Pinggy Tunnel
# --------------------------------------------------

print("\nStarting Pinggy Tunnel...")

tunnel = pinggy.start_tunnel(
    forwardto="127.0.0.1:11434",
    headermodification=[
        "u:Host:localhost:11434"
    ],
    httpsonly=True,
    autoreconnect=True
)

# --------------------------------------------------
# 4. Get public URL
# --------------------------------------------------

print("\n========================================")
print("Pinggy Public URLs:")
print("========================================")

print(tunnel.urls)

public_url = ""

if tunnel.urls:
    # Prefer HTTPS URL
    for url in tunnel.urls:
        if url.startswith("https://"):
            public_url = url.rstrip("/")
            break

    # Fallback
    if not public_url:
        public_url = tunnel.urls[0].rstrip("/")

if not public_url:
    raise RuntimeError("Pinggy did not provide a public URL.")

print("\nPublic URL:")
print(public_url)

# --------------------------------------------------
# 5. Verify the tunnel
# --------------------------------------------------

print("\nTesting Pinggy Tunnel...")

tunnel_working = False

for i in range(15):
    try:
        response = requests.get(
            f"{public_url}/api/tags",
            timeout=5
        )

        print(
            f"Attempt {i + 1}: "
            f"HTTP {response.status_code}"
        )

        if response.status_code == 200:
            print("\nTunnel is working cleanly!")
            print(response.text)
            tunnel_working = True
            break

        elif response.status_code in (400, 401, 403):
            print(
                f"\n[Error] Ollama rejected the request "
                f"(HTTP {response.status_code})."
            )
            print(
                "Check the Host header and Ollama configuration."
            )
            print("Response:", response.text)
            break

        else:
            print(
                f"[Warning] Unexpected HTTP status: "
                f"{response.status_code}"
            )
            print("Response:", response.text)

    except requests.exceptions.RequestException as e:
        print(
            f"Attempt {i + 1}: "
            f"Waiting for tunnel... "
            f"{type(e).__name__}: {e}"
        )

    time.sleep(2)

if not tunnel_working:
    raise RuntimeError(
        "Pinggy Tunnel is not forwarding correctly."
    )

# --------------------------------------------------
# 6. Keep everything alive
# --------------------------------------------------

print("\n========================================")
print("Ollama + Pinggy are running.")
print("========================================")
print("Public URL:", public_url)
print("\nDO NOT close this cell.")
print("Use the URL above from your local machine.")
```
